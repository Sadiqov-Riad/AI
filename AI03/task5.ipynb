```python
import pandas as pd
```

```python
data_tickets = {
    "Фильм": ["Начало", "Интерстеллар", "Начало", "Тенет", "Интерстеллар", "Дюна", "Дюна", "Тенет", "Начало", "Дюна", "Тенет", "Интерстеллар"],
    "Город": ["Москва", "Москва", "СПб", "Москва", "СПб", "Казань", "Москва", "СПб", "Казань", "СПб", "Казань", "Казань"],
    "Проданные билеты": [150, 200, 120, 80, 180, 300, 250, 90, 110, 210, 75, 160]
}

tickets_df = pd.DataFrame(data_tickets)
print("Таблица продаж билетов:")
display(tickets_df)
```

```python
# Общее количество проданных билетов по каждому фильму
sales_by_film = tickets_df.groupby("Фильм")["Проданные билеты"].sum()
print("Продажи по фильмам:")
print(sales_by_film)
```

```python
# Общее количество билетов по каждому городу
sales_by_city = tickets_df.groupby("Город")["Проданные билеты"].sum()
print("\nПродажи по городам:")
print(sales_by_city)
```

```python
# Фильм с наибольшим количеством продаж
top_film = sales_by_film.idxmax()
max_tickets = sales_by_film.max()
print(f"\nФильм с наибольшим количеством продаж: {top_film} ({max_tickets} билетов)")
```
