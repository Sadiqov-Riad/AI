```python
import pandas as pd

data = """Название,Автор,Год издания,Цена
Преступление и наказание,Фёдор Достоевский,1866,450
Мастер и Маргарита,Михаил Булгаков,1967,520
Искусство легких касаний,Виктор Пелевин,2019,650
1984,Джордж Оруэлл,1949,380
Внутри убийцы,Майк Омер,2018,490
Тревожные люди,Фредрик Бакман,2019,580
"""

with open('books.csv', 'w', encoding='utf-8') as f:
    f.write(data)

print('Файл books.csv создан.')
```

```python
df = pd.read_csv('books.csv')
print("Весь DataFrame:")
display(df)
```

```python
recent_books = df[df['Год издания'] > 2015]
print("\nКниги, изданные после 2015 года:")
display(recent_books)
```

```python
sorted_df = df.sort_values(by='Цена', ascending=True)
print("\nОтсортированный по цене DataFrame:")
display(sorted_df)
```
