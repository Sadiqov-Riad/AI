```python
import pandas as pd
```

```python
data_employees = {
    "Имя": ["Алексей", "Мария", "Алексей", "Иван", "Мария", "Иван", "Елена", "Елена"],
    "Проект": ["Альфа", "Альфа", "Бета", "Гамма", "Гамма", "Альфа", "Бета", "Гамма"],
    "Часы": [10, 15, 5, 20, 12, 8, 25, 10]
}

emp_df = pd.DataFrame(data_employees)
print("Исходный DataFrame:")
display(emp_df)
```

```python
# Общее количество часов по каждому сотруднику
hours_per_employee = emp_df.groupby("Имя")["Часы"].sum()
print("Часы по сотрудникам:")
print(hours_per_employee)
```

```python
# Общее количество часов по каждому проекту
hours_per_project = emp_df.groupby("Проект")["Часы"].sum()
print("Часы по проектам:")
print(hours_per_project)
```

```python
# Сотрудник, затративший больше всего часов
max_hours_employee = hours_per_employee.idxmax()
max_hours_value = hours_per_employee.max()
print(f"Больше всего часов ({max_hours_value}) затратил(а): {max_hours_employee}")
```
