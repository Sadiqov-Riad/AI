```python
import pandas as pd
```

```python
df = pd.read_csv("data/orders.csv")
```

```python
print("Первые десять строк:")
display(df.head(10))
```

```python

count_orders = df["Клиент"].value_counts()
print(count_orders)
```

```python
# Найдите макс и мин суммы заказов.

max_sum = df["Сумма"].max()
min_sum = df["Сумма"].min()

print(max_sum)
print(min_sum)
```

```python
# Подсчет общей сумму всех заказов.

total_sum = df["Сумма"].sum()

print(total_sum)
```
