```python
import pandas as pd
```

```python
data_food = {
    "Продукт": ["Курица", "Лосось", "Орехи", "Яблоко", "Банан", "Шоколад", "Творог", "Сыр", "Огурец", "Говядина"],
    "Категория": ["Мясо", "Рыба", "Снеки", "Фрукты", "Фрукты", "Снеки", "Молочные", "Молочные", "Овощи", "Мясо"],
    "Калории": [165, 208, 600, 52, 89, 540, 150, 400, 15, 250],
    "Белки": [31, 20, 15, 0.3, 1.1, 5, 18, 25, 0.8, 26]
}

food_df = pd.DataFrame(data_food)
display(food_df)
```

```python
# Продукты с калорийностью выше 300
high_cal = food_df[food_df["Калории"] > 300]
print("Продукты с калорийностью > 300:")
display(high_cal)
```

```python
# Среднее количество белков по категориям
avg_protein = food_df.groupby("Категория")["Белки"].mean()
print("Среднее количество белков по категориям:")
print(avg_protein)
```

```python
# Сортировка по калорийности в порядке убывания
sorted_food = food_df.sort_values(by="Калории", ascending=False)
print("Отсортированный DataFrame:")
display(sorted_food)
```
